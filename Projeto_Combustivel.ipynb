```python
import pandas as pd # pd é o nome da biblioteca
                    # pandas é uma biblioteca dopython para analise de dados,...
                    # ... maniupulação de tabelas, limpeza de dados e leitura de arquivos (csv, excel)

import numpy as np # numpy é uma biblioteca para computação numérica

# .ipynb serve para rodar os códigos separados por células, é o formato do jupyter notebook
# sempre que ligar o pc novamente, rodar todos os códigos no "run all"
```

```python
dados = "https://www.ispdados.rj.gov.br/Arquivos/BaseDPEvolucaoMensalCisp.csv"
df = pd.read_csv(dados, encoding= "latin1", sep= ";") # df = dataframe (estrutura de dados em tabelas)
                                                      # read_csv() faz ler o arquivo em formato de tabelas
                                                      # pd é o nome da biblioteca utilizada
                                                      # dados é para citar a variavel utilizada
                                                      # encoding= é para "traduzir" para a versão do windows
                                                      # sep= ";" informa o separador do arquivo

display(df.head()) # display() mostra em formato de tabela, mas não tem a necessidade de escrever
                   # df.head() mostra só as primeiras 5 (ou a quantidade que colocar entre os parenteses) linhas de um dataframe (tabela) 
                   # df.tail() mostra as 5 ultimas linhas
```

```python
df_furto_celular_cisp = df.groupby('cisp')['furto_celular'].sum().reset_index() # .groupby é para agrupar o que for mencionado (cisp)
                                                                                # (cisp) executa o agroupby
                                                                                # [furto_celular] mostra a coluna que será mencionada
                                                                                # sum() para somar o que foi mencionado
                                                                                # .reset_index() é para refazer uma nova tabela mencionado no df 
                                                                                # ... no início do código
                                                                                # cisp é uma das colunas do arquivo mencionado

df_furto_celular_cisp       # df_furto_celular_cisp é para mostrar a tabela após os códigos acima                                                             
```

```python
dados = np.array([12, 15, 17, 20, 22, 25, 28, 30, 35, 40]) # Criei uma lista, porém, poderia ser uma já existente
                                                           # array é uma estrutura que guarda varios elementos iguais (q1 q2 q3)
q1 = np.percentile(dados, 25) # 25% da lista # .percentile() coloca o em porcentagem
q2 = np.percentile(dados, 50) # 50% da lista # .percentile() coloca o em porcentagem
q3 = np.percentile(dados, 75) # 75% da lista # .percentile() coloca o em porcentagem
print(f'Primeiro quartril (Q1): {q1}')
print(f'Segundo quartril (Q2): {q2}')
print(f'Terceiro quartril (Q3): {q3}')
```

```python
array_furto_celular = np.array(df_furto_celular_cisp['furto_celular'])
q1_furto_celular = np.percentile(array_furto_celular, 25) # array transforma uma lista voltada para cálculos específicos
q2_furto_celular = np.percentile(array_furto_celular, 50)
q3_furto_celular = np.percentile(array_furto_celular, 75)
media = np.mean(array_furto_celular) 
mediana = np.median(array_furto_celular)
distancia = (media - mediana) / mediana
iqr = q3_furto_celular - q1_furto_celular
limite_superior = q3_furto_celular + (1.5 * iqr)
limite_inferior = q1_furto_celular - (1.5 * iqr)


print(f'Primeiro quartril (Q1): {q1_furto_celular}')
print(f'Segundo quartril (Q2): {q2_furto_celular}')
print(f'Terceiro quartril (Q3): {q3_furto_celular}')
print(f'O valor da média é {media}')
print(f'O valor da mediana é {mediana}')
print(f'A distância entre média e mediana é {distancia}')
print(f'O limite superior é de {limite_superior}')
print(f'O limite inferior é de {limite_inferior}')
```

```python
# Identificar o que puxa a media para cima (outiliers)

outiliers_cisp_furto_celular = df_furto_celular_cisp.loc[df_furto_celular_cisp['furto_celular'] >= limite_superior]
outiliers_cisp_furto_celular = outiliers_cisp_furto_celular.sort_values(by='furto_celular', ascending=False)
outiliers_cisp_furto_celular
```

```python
df_menor_q1_furto_celular = df_furto_celular_cisp.loc[df_furto_celular_cisp['furto_celular'] < q1_furto_celular]

df_menor_q1_furto_celular = df_menor_q1_furto_celular.sort_values(by='furto_celular', ascending= False) # ascending= False é para colocar do maior para o menor
                                                                                                        # by= '   ' indica por qual coluna o usuário quer ordenar

df_menor_q1_furto_celular
```

```python
df_ocorrencias_veiculos = df.groupby('cisp')[['roubo_veiculo', 'furto_veiculos']].sum().reset_index()
df_ocorrencias_veiculos = df_ocorrencias_veiculos.sort_values(by='roubo_veiculo')
df_ocorrencias_veiculos
```

```python
df_ocorrencias_veiculos['total_ocorrencias'] = df_ocorrencias_veiculos['roubo_veiculo'] + df_ocorrencias_veiculos['furto_veiculos']
df_ocorrencias_veiculos
```

```python
array_ocorrencia = np.array(df_ocorrencias_veiculos[['total_ocorrencias']])
q1_roubo_veiculos = np.percentile(array_ocorrencia, 25)
q2_roubo_veiculos = np.percentile(array_ocorrencia, 50)
q3_roubo_veiculos = np.percentile(array_ocorrencia, 75)
media_ocorrencia = np.mean(array_ocorrencia)
mediana_ocorrencia = np.median(array_ocorrencia)
distancia_ocorrencia = (media_ocorrencia - mediana_ocorrencia) / mediana_ocorrencia
iqr = q3_furto_celular - q1_furto_celular
limite_superior = q3_furto_celular + (1.5 * iqr)
limite_inferior = q1_furto_celular - (1.5 * iqr)

print(f'O terceiro quatril é de {q3_roubo_veiculos}')
print(f'O limite superior é de {limite_superior}')
print(f'O limite inferior é de {limite_inferior}')
```

```python
outiliers_cisp_ocorrencias_veiculo = df_ocorrencias_veiculos.loc[df_ocorrencias_veiculos['total_ocorrencias'] >= limite_superior]
outiliers_cisp_ocorrencias_veiculo = outiliers_cisp_ocorrencias_veiculo.sort_values(by='total_ocorrencias', ascending=False)
outiliers_cisp_ocorrencias_veiculo
```
